merge() — combine using a common column

```python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 70]
})
```

```python
result = pd.merge(students, marks, on="ID")
```

```python
print(result)
```

concat() → stack/combine DataFrames.
