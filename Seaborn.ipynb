```python
import pandas as pd

data = {
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108],

    "Product": [
        "Laptop", "Mouse", "T-Shirt", "Headphones",
        "Jeans", "Keyboard", "Shoes", "Monitor"
    ],

    "Category": [
        "Electronics", "Electronics", "Clothing", "Electronics",
        "Clothing", "Electronics", "Clothing", "Electronics"
    ],

    "Price": [120000, 2500, 3000, 8000, 5000, 4500, 7000, 35000],

    "Quantity": [1, 2, 3, 1, 2, 1, 2, 1],

    "City": [
        "Peshawar", "Lahore", "Karachi", "Peshawar",
        "Lahore", "Karachi", "Peshawar", "Lahore"
    ]
}

df = pd.DataFrame(data)

df["Revenue"] = df["Price"] * df["Quantity"]

print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sns.scatterplot(data = df, x = "Price", y = "Quantity")
plt.show()
```

```python
sns.scatterplot(data = df, x = "Price", y = "Quantity", hue = "Category")   # hue = "column_name"  (when you want another variable to distinguish groups)
plt.show()
```

```python
sns.lineplot(data=df, x="Price", y="Quantity")

plt.title("Price vs Quantity")
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [12000, 15000, 11000, 18000, 22000]
```

```python
sns.lineplot(x=days, y=sales, marker="o")

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
```

sns.barplot() → compares categories using an aggregate, mean by default.

```python
sns.barplot(
    data=df,
    x="Category",
    y="Price"
)

plt.title("Average Price by Category")
plt.show()
```

# Difference between matplotlib and seaborn in case of bar chart
Matplotlib:
"Here are my values → draw them."

Seaborn barplot:
"Here is my raw data → calculate the average for each category → draw it."

```python
sns.histplot(data=df, x="Price")

plt.title("Distribution of Prices")
plt.show()
```

kde=True adds a smooth curve showing the shape of the distribution.

```python
sns.histplot(data=df, x="Price", kde=True)

plt.title("Distribution of Prices")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

correlation = df[["Price", "Quantity", "Revenue"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(
    df,
    vars=["Price", "Quantity", "Revenue"],
    hue="Category"
)

plt.show()
```
