# **Matplotlib**

```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd

data = {
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Product": [
        "Laptop", "Mouse", "T-Shirt", "Headphones",
        "Jeans", "Keyboard", "Shoes", "Monitor"
    ],
    "Category": [
        "Electronics", "Electronics", "Clothing", "Electronics",
        "Clothing", "Electronics", "Clothing", "Electronics"
    ],
    "Price": [120000, 2500, 3000, 8000, 5000, 4500, 7000, 35000],
    "Quantity": [1, 2, 3, 1, 2, 1, 2, 1],
    "City": [
        "Peshawar", "Lahore", "Karachi", "Peshawar",
        "Lahore", "Karachi", "Peshawar", "Lahore"
    ]
}

df = pd.DataFrame(data)

df["Revenue"] = df["Price"] * df["Quantity"]
```

For python → ML path, you mainly need:

# Bar chart

# Line chart

# Scatter plot

# Histogram

# Pie chart — basic knowledge

# Box plot — useful for ML/data analysis

# Heatmap — usually through Seaborn later

Categories → Compare → Bar

Time/order → Trend/change → Line

Number vs Number → Relationship → Scatter

One numerical variable → Distribution → Histogram

What percentage/share does each category contribute to the total? -->  pie chart

Distribution + spread + outliers → Box plot



```python
print(df[["Product", "Revenue"]])
```

**Bar chart**

```python
plt.bar(df["Product"], df["Revenue"])
plt.show()
```

**Line chart**

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [12000, 15000, 11000, 18000, 22000]

plt.plot(days, sales , marker = "o")

plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
```

**Scatter chart**

```python
plt.scatter(df["Price"], df["Quantity"])

plt.title("Price vs Quantity")
plt.xlabel("Price")
plt.ylabel("Quantity")

plt.show()
```

**Histogram**

```python
plt.hist(df["Price"])

plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

**Boxplot**

```python
plt.boxplot(df["Price"])

plt.title("Distribution of Product Prices")
plt.ylabel("Price")

plt.show()
```

**Pie-Chart**

```python
category_revenue = df.groupby("Category")["Revenue"].sum()
```

```python
plt.pie(
    category_revenue,
    labels=category_revenue.index,
    autopct="%1.1f%%"
)

plt.title("Revenue by Category")
plt.show()
```

**Title , x-y axis , rotation and figure size**

```python
plt.figure(figsize=(5, 3))
plt.bar(df["Product"], df["Revenue"])

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.show()
```

| Your question                                  | Chart            |
| ---------------------------------------------- | ---------------- |
| Compare different categories                   | 📊 **Bar**       |
| See change over time                           | 📈 **Line**      |
| See relationship between 2 numerical variables | 🔵 **Scatter**   |
| See distribution of numerical values           | 📊 **Histogram** |
| See parts/percentages of one whole             | 🥧 **Pie**       |
| Find spread/outliers                           | 📦 **Box plot**  |
| See correlation between many variables         | 🔥 **Heatmap**   |


**Naming the bars with their height**

```python
plt.figure(figsize=(5, 4))

bars = plt.bar(df["Product"], df["Revenue"])

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(bar.get_height()),
        ha="center",
        va="bottom"
    )

plt.show()
```

**Subplot**

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.bar(df["Product"], df["Revenue"])
plt.title("Revenue by Product")
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
plt.plot(days, sales, marker="o")
plt.title("Daily Sales")

plt.tight_layout()
plt.show()
```
